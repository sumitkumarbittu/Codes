```python
import matplotlib.pyplot as plt

# Data
x = [1, 2, 3, 4]
y = [10, 20, 25, 30]

# Create a line plot
plt.plot(x, y)
plt.title("Basic Line Plot")  # Add a title
plt.xlabel("X-axis Label")    # Label the X-axis
plt.ylabel("Y-axis Label")    # Label the Y-axis
plt.show()
```

```python
# Create a figure and axes
fig, ax = plt.subplots()
ax.plot(x, y, label="Line 1")  # Plot on the axes
ax.set_title("Figure with Axes")  # Add title
ax.set_xlabel("X-axis")           # Set X-axis label
ax.set_ylabel("Y-axis")           # Set Y-axis label
ax.legend()                       # Add a legend
plt.show()
```

```python
# Create a 1x2 grid of subplots
fig, axs = plt.subplots(1, 2, figsize=(10, 4))

# First subplot
axs[0].plot(x, y, color='blue', label="Line 1")
axs[0].set_title("First Plot")
axs[0].legend()

# Second subplot
axs[1].plot(x, [15, 25, 35, 45], color='red', label="Line 2")
axs[1].set_title("Second Plot")
axs[1].legend()

plt.tight_layout()  # Adjust layout to avoid overlap
plt.show()
```

```python
fig, ax = plt.subplots()
ax.plot(x, y, label="Blue Line", color="blue", linestyle="--")
ax.plot(x, [15, 25, 35, 45], label="Red Line", color="red", linestyle="-")
ax.set_title("Customized Plot")
ax.set_xlabel("X-axis")
ax.set_ylabel("Y-axis")
ax.grid(True)        # Add grid lines
ax.legend()          # Add legend
plt.show()
```

```python
# Use a built-in style
plt.style.use("ggplot")

# Create a simple plot
plt.plot(x, y, marker='o', linestyle='-', color='green')
plt.title("Styled Plot")
plt.show()
```

```python
plt.scatter(x, y, color='purple')
plt.title("Scatter Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

```python
categories = ["A", "B", "C", "D"]
values = [3, 7, 8, 5]

plt.bar(categories, values, color='orange')
plt.title("Bar Plot")
plt.xlabel("Categories")
plt.ylabel("Values")
plt.show()
```

```python
data = [1, 2, 2, 3, 3, 3, 4, 4, 5]
plt.hist(data, bins=5, color='skyblue', edgecolor='black')
plt.title("Histogram")
plt.xlabel("Bins")
plt.ylabel("Frequency")
plt.show()
```

```python
sizes = [15, 30, 45, 10]
labels = ["A", "B", "C", "D"]
plt.pie(sizes, labels=labels, autopct='%1.1f%%', startangle=90)
plt.title("Pie Chart")
plt.show()
```

```python
plt.plot(x, y)
plt.title("Save Example")
plt.savefig("plot.png")  # Save the figure
plt.show()
```
