```python
import pandas as pd

# Creating a DataFrame from a dictionary
data = {
    "Name": ["Alice", "Bob", "Charlie"],
    "Age": [25, 30, 35],
    "City": ["New York", "Los Angeles", "Chicago"]
}
df = pd.DataFrame(data)
print(df)
```

```python
df["Country"] = "USA"  # Add a new column
print(df)

```

```python
df = df.drop("City", axis = 1)  # Remove the column
print(df)
```

```python
print(df["Country"])       # Select a single column
```

```python
print(df[["Name", "Age"]])  # Select multiple columns
```

```python
print(df.iloc[0])       # Select row by index (position)
```

```python

print(df.loc[0])        # Select row by index (label)
```

```python

print(df.loc[df["Age"] > 30])  # Rows where Age > 30
```

```python
df.set_index("Name", inplace=True)  # Set 'Name' column as index
print(df)
```

```python

df.reset_index(inplace=True)  # Reset index
print(df)
```

```python
df1 = pd.DataFrame({"A": [1, 2], "B": [3, 4]})
df2 = pd.DataFrame({"A": [5, 6], "B": [7, 8]})

```

```python
# Concatenate along rows
result = pd.concat([df1, df2], axis=0)
print(result)
```

```python
# Concatenate along columns
result = pd.concat([df1, df2], axis=1)
print(result)
```

```python
df1 = pd.DataFrame({"ID": [1, 2], "Name": ["Alice", "Bob"]})
df2 = pd.DataFrame({"ID": [1, 2], "Age": [25, 30]})
```

```python

# Inner join (default)
merged = pd.merge(df1, df2, on="ID")
print(merged)
```

```python

# Left join
merged = pd.merge(df1, df2, on="ID", how="left")
print(merged)
```

```python
data = {
    "Department": ["HR", "HR", "IT", "IT", "Finance"],
    "Employee": ["Alice", "Bob", "Charlie", "David", "Eva"],
    "Salary": [50000, 60000, 70000, 80000, 90000]
}
df = pd.DataFrame(data)


```

```python
# Group by Department and calculate the mean salary
grouped = df.groupby("Department")["Salary"].mean()
print(grouped)
```

```python

# Multiple aggregation functions
grouped = df.groupby("Department")["Salary"].agg(["mean", "max"])
print(grouped)
```

```python
# Group by Department and count the number of employees
grouped = df.groupby("Department").size()
print(grouped)

```

```python
# Custom aggregation
grouped = df.groupby("Department")["Salary"].apply(lambda x: x.sum())
print(grouped)
```

```python
data = {
    "Department": ["HR", "HR", "IT", "IT"],
    "Employee": ["Alice", "Bob", "Charlie", "David"],
    "Salary": [50000, 60000, 70000, 80000]
}
df = pd.DataFrame(data)
```

```python

pivot = df.pivot_table(values="Salary", index="Department", aggfunc="mean")
print(pivot)
```

```python
melted = pd.melt(df, id_vars=["Department"], value_vars=["Salary"])
print(melted)
```
