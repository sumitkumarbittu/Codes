```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Load built-in dataset
tips = sns.load_dataset("tips")
print(tips.head())
```

```python
sns.scatterplot(data=tips, x="total_bill", y="tip", hue="time")
plt.title("Scatter Plot of Total Bill vs Tip")
plt.show()
```

```python
sns.lineplot(data=tips, x="size", y="total_bill")
plt.title("Line Plot of Size vs Total Bill")
plt.show()
```

```python
sns.barplot(data=tips, x="day", y="total_bill", ci="sd", hue="sex")
plt.title("Bar Plot of Total Bill by Day")
plt.show()
```

```python
sns.boxplot(data=tips, x="day", y="total_bill", hue="sex")
plt.title("Box Plot of Total Bill by Day")
plt.show()
```

```python
sns.violinplot(data=tips, x="day", y="total_bill", hue="sex", split=True)
plt.title("Violin Plot of Total Bill by Day")
plt.show()
```

```python
sns.histplot(data=tips, x="total_bill", kde=True, bins=20)
plt.title("Histogram of Total Bill")
plt.show()
```

```python
sns.kdeplot(data=tips, x="total_bill", hue="sex", fill=True)
plt.title("KDE Plot of Total Bill")
plt.show()
```

```python
sns.pairplot(tips, hue="sex")
plt.show()
```

```python
corr = tips.corr()
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

```python
sns.set_theme(style="darkgrid")  # Other styles: "whitegrid", "dark", "white", "ticks"
sns.boxplot(data=tips, x="day", y="total_bill")
plt.title("Box Plot with Darkgrid Theme")
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.barplot(data=tips, x="day", y="total_bill")
plt.title("Larger Plot")
plt.show()
```

```python
sns.scatterplot(data=tips, x="total_bill", y="tip", hue="time")
plt.title("Scatter Plot of Total Bill vs Tip")
plt.xlabel("Total Bill ($)")
plt.ylabel("Tip ($)")
plt.show()
```

```python
g = sns.FacetGrid(tips, col="time", row="sex", margin_titles=True)
g.map(sns.scatterplot, "total_bill", "tip")
plt.show()
```
